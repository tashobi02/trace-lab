# Artifact Integrity

This library notebook provides functions to verify the SHA256 hashes and identities of original evidence files.


```
import json
import hashlib
from pathlib import Path

def verify_file_hash(filepath, expected_hash):
    """Verify that a file matches its expected SHA-256 hash."""
    sha256 = hashlib.sha256()
    with open(filepath, "rb") as f:
        for chunk in iter(lambda: f.read(4096), b""):
            sha256.update(chunk)
    return sha256.hexdigest() == expected_hash

print("Artifact Integrity functions loaded.")
```
